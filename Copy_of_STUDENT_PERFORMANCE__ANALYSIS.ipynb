```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
data = {
    'Student_Name': ['suchi', 'Rishta', 'sujith', 'chatu', 'Moni',
                     'Teja', 'Dhana', 'Bhavi', 'Reena', 'Revanth'],

    'Maths': [85, 70, 92, 65, 88, 76, 95, 60, 82, 78],

    'Science': [90, 75, 89, 68, 92, 80, 94, 65, 85, 79],

    'English': [88, 72, 95, 70, 86, 78, 91, 62, 89, 81],

    'Attendance': [95, 80, 98, 70, 92, 85, 99, 65, 90, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Total'] = df['Maths'] + df['Science'] + df['English']

print(df[['Student_Name', 'Total']])
```

```python
df['Average'] = df['Total'] / 3

print(df[['Student_Name', 'Average']])
```

```python
topper = df.loc[df['Total'].idxmax()]

print("Topper:", topper['Student_Name'])
print("Total Marks:", topper['Total'])
```

```python
lowest = df.loc[df['Total'].idxmin()]

print("Lowest Scorer:", lowest['Student_Name'])
print("Total Marks:", lowest['Total'])
```

```python
class_average = df['Average'].mean()

print("Class Average:", class_average)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df['Student_Name'], df['Total'])

plt.title('Student Performance Analysis')
plt.xlabel('Student Names')
plt.ylabel('Total Marks')

plt.show()
```

```python
subjects = ['Maths', 'Science', 'English']

averages = [
    df['Maths'].mean(),
    df['Science'].mean(),
    df['English'].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(subjects, averages)

plt.title('Subject-wise Average Performance')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Attendance'], df['Average'])

plt.title('Attendance vs Student Performance')
plt.xlabel('Attendance Percentage')
plt.ylabel('Average Marks')

plt.show()
```

```python
correlation = df['Attendance'].corr(df['Average'])

print("Attendance and Marks Correlation:", correlation)
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[['Maths', 'Science', 'English', 'Attendance', 'Average']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Student Performance Correlation Heatmap')
plt.show()
```
